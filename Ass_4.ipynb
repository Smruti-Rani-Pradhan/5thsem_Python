```python
#3
list1 = []  

while True:
    print("\n1. Create  2. Display  3. Insert  4. Delete  5. Exit\n")
    ch = int(input("Enter your choice: "))

    if ch == 1:
        size = int(input("Enter size: "))
        list1 = [int(input("Enter number: ")) for _ in range(size)]

    elif ch == 2:
        print("List:", list1)

    elif ch == 3:
        pos = int(input("Enter position: "))
        val = int(input("Enter value: "))
        if 0 <= pos <= len(list1):
            list1.insert(pos, val)
            print("Updated list:", list1)
        else:
            print("Invalid position!")

    elif ch == 4:
        pos = int(input("Enter position: "))
        if 0 <= pos < len(list1):
            list1.pop(pos)
            print("Updated list:", list1)
        else:
            print("Invalid position!")

    elif ch == 5:
        print("Exiting program...")
        break

    else:
        print("Invalid choice! Please enter 1–5.")
```
