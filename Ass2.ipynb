```python
#6
n=int(input("Enter a number:"))
c=0
for i in range(2,int(n/2)):
    if n==1:
        print("Not a prime Number")
if n%i==0:
    c+=1;
if c==0:
    print("Prime Number")
else:
    print("Not a Prime Number")    
```
